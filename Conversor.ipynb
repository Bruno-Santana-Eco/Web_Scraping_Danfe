```python
import os
from brazilfiscalreport.danfe import Danfe

# Caminho para a pasta com os arquivos XML
input_folder = r'C:\Users\bruno.bezerra\Desktop\Ambiente de Trabalho\Web Scraping\Download XML'

# Caminho para a pasta onde os PDFs serão salvos
output_folder = r'C:\Users\bruno.bezerra\Desktop\Ambiente de Trabalho\Web Scraping\PDF Convertido'

# Certifique-se de que a pasta de saída exista
os.makedirs(output_folder, exist_ok=True)

# Iterar sobre todos os arquivos na pasta de entrada
for file_name in os.listdir(input_folder):
    if file_name.endswith('.xml'):
        # Caminho completo para o arquivo XML
        xml_file_path = os.path.join(input_folder, file_name)
        
        # Carregar o conteúdo do arquivo XML
        with open(xml_file_path, "r", encoding="utf8") as file:
            xml_content = file.read()

        # Instanciar o objeto DANFE com o conteúdo XML
        danfe = Danfe(xml=xml_content)

        # Gerar o nome do arquivo PDF correspondente
        pdf_file_name = f"{os.path.splitext(file_name)[0]}.pdf"
        pdf_output_path = os.path.join(output_folder, pdf_file_name)

        # Salvar o PDF gerado
        danfe.output(pdf_output_path)

        print(f"PDF gerado com sucesso: {pdf_output_path}")
```
