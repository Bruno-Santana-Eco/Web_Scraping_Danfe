```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# Configuração do caminho para o arquivo Excel
excel_path = 'C:/Users/bruno.bezerra/Desktop/Ambiente de Trabalho\Web Scraping/NF não escrituradas.xlsx'

# Carregar o arquivo Excel e extrair a coluna 'Chave de Acesso'
df = pd.read_excel(excel_path, sheet_name='Planilha1')
df = df.iloc[5:, :].reset_index(drop=True)
df.columns = df.iloc[0]
df = df.iloc[1:, :].reset_index(drop=True)
df_chaves = df[['Chave de Acesso']]  # Filtrar a coluna 'Chave de Acesso'
chaves_danfe = df_chaves['Chave de Acesso'].tolist()  # Converter para lista

# Definir as opções do Chrome para permitir o download automático
chrome_options = webdriver.ChromeOptions()
chrome_options.add_experimental_option("prefs", {
    "download.default_directory": r"C:\Users\bruno.bezerra\Desktop\Ambiente de Trabalho\Web Scraping\Download XML",  # Definir o diretório de download
    "download.prompt_for_download": False,  # Desabilitar o prompt de download
    "download.directory_upgrade": True,  # Atualizar o diretório, se necessário
    "safebrowsing.enabled": True,  # Desativar verificações de segurança do Safe Browsing
    "safebrowsing.disable_download_protection": True,  # Desabilitar a proteção de downloads não verificados
    "profile.default_content_setting_values.automatic_downloads": 1  # Permitir downloads automáticos
})

# Caminho para o driver
driver_path = "C:/Users/bruno.bezerra/Desktop/Ambiente de Trabalho/Web Scraping/chromedriver-win64/chromedriver-win64/chromedriver.exe"

# Inicializar o driver com as opções configuradas
service = Service(driver_path)
driver = webdriver.Chrome(service=service, options=chrome_options)

# Acessar o site
driver.get("https://consultadanfe.com/?gclid=EAIaIQobChMI2Lahqfj9-gIVAkVIAB3McQr_EAAYASAAEgI8XvD_BwE#")
time.sleep(5)

# Fechar o pop-up (se existir)
try:
    fechar_botao = driver.find_element(By.XPATH, '/html/body/div[1]/div[5]/div/div/div/div/div/div/div/a[1]')
    fechar_botao.click()
    time.sleep(5)
except:
    print("Nenhum pop-up encontrado.")

# Loop para processar cada chave DANFE
for chave_danfe in chaves_danfe:
    print(f"Processando chave: {chave_danfe}")
    
    # Inserir a chave DANFE
    campo_chave = driver.find_element(By.XPATH, '/html/body/div[1]/div[2]/div/div[1]/div/div[1]/form[1]/input')
    campo_chave.clear()  # Limpar o campo antes de inserir a nova chave
    campo_chave.send_keys(chave_danfe)
    
    time.sleep(5)
    
    # Esperar o botão de buscar estar visível e clicar nele
    botao_buscar = WebDriverWait(driver, 2).until(
        EC.visibility_of_element_located((By.CSS_SELECTOR, "#form_one > button"))
    )
    driver.execute_script("arguments[0].scrollIntoView();", botao_buscar)
    driver.execute_script("arguments[0].click();", botao_buscar)
    
    # Loop para verificar se o link de download está disponível
    max_attempts = 30  # Máximo de tentativas (exemplo: 30 tentativas de 2 segundos, total de 1 minuto)
    attempts = 0
    download_iniciado = False
    
    while attempts < max_attempts:
        try:
            download_link = driver.find_element(By.XPATH, '/html/body/div[1]/div[4]/div/div/div[2]/div/div[3]/p[2]/a')
            download_link.click()
            download_iniciado = True
            print(f"Download iniciado para a chave: {chave_danfe}")
            break  # Sair do loop se o download for iniciado com sucesso
        except Exception as e:
            attempts += 1
            print(f"Tentativa {attempts} falhou. Tentando novamente em 2 segundos...")
            time.sleep(2)  # Espera 2 segundos antes de tentar novamente

    if not download_iniciado:
        print(f"Erro: Não foi possível iniciar o download para a chave {chave_danfe} após {attempts} tentativas.")
    
    time.sleep(2)

    # Fechar a janela de visualização para processar a próxima chave
    botao_fechar = WebDriverWait(driver, 10).until(
        EC.visibility_of_element_located((By.XPATH, "/html/body/div[1]/div[4]/div/div/div[1]/button/span"))
    )
    botao_fechar.click()    
    
    time.sleep(2)  # Esperar antes de processar a próxima chave


# Fechar o navegador após o loop
driver.quit()
```
